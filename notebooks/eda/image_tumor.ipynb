# **EDA of Lung-PET-CT-Dx | Images and Tumor Data**

Exploratory Data Analysis of Images and Tumor Data

## Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from src.data.dicom_metadata import build_metadata_dataframe

# Mismo tema y paleta que clinical_data.ipynb.
sns.set_theme(style='whitegrid', palette='Set2')
COLORS = sns.color_palette('Set2')
pd.set_option('display.max_columns', 100)

# Where the figures of this notebook are stored.
FIG_DIR = Path('../../results/reports/EDA/image_metadata')
FIG_DIR.mkdir(parents=True, exist_ok=True)

def savefig(name):
    plt.tight_layout()
    plt.savefig(FIG_DIR / name, dpi=150, bbox_inches='tight')
```

# Load metadata

```python
df = build_metadata_dataframe(
    root='../../data/raw/lung-pet-ct/images',
    cache_path='../../data/processed/image_metadata.parquet',
)
print(df.shape)
df.head()
```

```python
# Any unreadable / corrupt files?
n_err = df['read_error'].notna().sum()
print(f'Instances with read errors: {n_err}')
if n_err:
    display(df.loc[df['read_error'].notna(), ['file_path', 'read_error']].head())
```

# Dataset overview

```python
overview = {
    'instances (.dcm)': len(df),
    'patients': df['patient_folder'].nunique(),
    'studies': df['study_uid'].nunique(),
    'series': df['series_uid'].nunique(),
}
pd.Series(overview, name='count').to_frame()
```

```python
# Instances per modality.
mod_counts = df['modality'].value_counts(dropna=False)
display(mod_counts.to_frame('instances'))

# Series per modality (a series is the natural acquisition unit).
series_mod = df.groupby('series_uid')['modality'].first().value_counts(dropna=False)
display(series_mod.to_frame('series'))
```

```python
pat_mod = df.groupby('patient_folder')['modality'].agg(lambda s: '+'.join(sorted(set(s.dropna()))))
pat_mod.value_counts().to_frame('patients')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Izquierda: cuántas instancias hay de cada modalidad (CT, PT).
mod_counts.index = mod_counts.index.str.replace('CT', 'TAC').str.replace('PT', 'PET')
mod_counts.plot.bar(ax=axes[0], color=COLORS[0])
axes[0].set_title('Imágenes por modalidad')
axes[0].set_ylabel('instancias')

# Derecha: cuántos pacientes tienen cada combinación de modalidades.

pat_mod = pat_mod.str.replace('CT+PT', 'TAC+PET').str.replace('CT', 'TAC')
pat_mod.value_counts().plot.bar(ax=axes[1], color=COLORS[1])
axes[1].set_title('Pacientes por combinación de modalidades')
axes[1].set_ylabel('pacientes')

for ax in axes:
    ax.tick_params(axis='x', rotation=0)
    ax.bar_label(ax.containers[0], fmt='%d', padding=3)
    ax.margins(y=0.15)

savefig('modality_counts.png')
plt.show()
```

## Distribución de los pacientes por estudio y número de imágenes por paciente.

```python
# Estudios por paciente: un estudio es una visita/adquisición completa.
estudios_por_paciente = df.groupby('patient_folder')['study_uid'].nunique()
conteo_estudios = estudios_por_paciente.value_counts().sort_index()

# Imágenes por paciente, separadas según si el paciente tiene PET o no: la
# distribución es bimodal y el motivo es justamente ese.
imagenes_por_paciente = df.groupby('patient_folder').size()
tiene_pet = df.groupby('patient_folder')['modality'].apply(lambda s: 'PT' in set(s))
grupos = tiene_pet.map({False: 'Solo TAC', True: 'TAC+PET'})

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

conteo_estudios.plot.bar(ax=axes[0], color=COLORS[0])
axes[0].set_title('Pacientes por número de estudios')
axes[0].set_xlabel('estudios del paciente')
axes[0].set_ylabel('pacientes')
axes[0].bar_label(axes[0].containers[0], fmt='%d', padding=3)
axes[0].tick_params(axis='x', rotation=0)
axes[0].margins(y=0.15)

bins = np.linspace(0, imagenes_por_paciente.max(), 40)
for etiqueta, color in [('Solo TAC', COLORS[0]), ('TAC+PET', COLORS[1])]:
    sub = imagenes_por_paciente[grupos == etiqueta]
    axes[1].hist(sub, bins=bins, color=color, alpha=0.75,
                 label=f'{etiqueta} (n={len(sub)}, mediana {sub.median():.0f})')
axes[1].set_title('Imágenes por paciente')
axes[1].set_xlabel('imágenes del paciente')
axes[1].set_ylabel('pacientes')
axes[1].legend(fontsize=9)

savefig('patients_studies_images.png')
plt.show()

resumen = pd.DataFrame({
    'estudios': estudios_por_paciente.describe(),
    'imágenes': imagenes_por_paciente.describe(),
}).round(1)
display(resumen)
display(imagenes_por_paciente.groupby(grupos).describe()[['count', 'min', '50%', 'max']].round(0))
```

## Cohort sub-groups
Patients are coded `Lung_Dx-<A|B|E|G>NNNN`. The letter encodes the
diagnostic sub-cohort.

```python
cohort = df.drop_duplicates('patient_folder')['patient_folder'].str.extract(r'Lung_Dx-([A-Z])')[0]
cohort.value_counts().sort_index().to_frame('patients')
```

```python
# La letra del código de paciente codifica el subtipo histológico.
HISTOLOGY = {
    'A': 'Adenocarcinoma',
    'B': 'Small Cell Carcinoma',
    'E': 'Large Cell Carcinoma',
    'G': 'Squamous Cell Carcinoma',
}

# cohort es una letra por paciente, no un recuento: hay que contar antes de dibujar.
counts = cohort.value_counts().rename(index=HISTOLOGY).sort_values()
pct = 100 * counts / counts.sum()

# Barras horizontales: los nombres histológicos no caben girados bajo un eje x.
fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(counts.index, counts.to_numpy(), color=COLORS[0], zorder=3)
ax.bar_label(ax.containers[0],
             labels=[f'{n}  ({p:.1f}%)' for n, p in zip(counts, pct)],
             padding=4, fontsize=9.5)

ax.set_xlabel('pacientes')
ax.set_title(f'Subtipo histológico (n = {counts.sum()} pacientes)')
ax.margins(x=0.16)
ax.grid(axis='y', visible=False)

savefig('cancer_type_counts.png')
plt.show()

counts.sort_values(ascending=False).to_frame('patients').assign(
    pct=pct.sort_values(ascending=False).round(1))
```
